```python
import pandas as pd

df = pd.read_csv('/kaggle/input/youtube-new/INvideos.csv')

print("Dataset loaded successfully")
df.head()
```

```python
print("Rows and columns:", df.shape)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()

df['trending_date'] = pd.to_datetime(df['trending_date'], errors='coerce')

df['publish_time'] = pd.to_datetime(df['publish_time'], errors='coerce')

df['title'] = df['title'].fillna('Unknown')

df['channel_title'] = df['channel_title'].fillna('Unknown')

df['tags'] = df['tags'].fillna('')

df = df.dropna(subset=['video_id'])

print("Cleaning completed")
print("New shape:", df.shape)
```

```python
top_views = df.sort_values('views', ascending=False)

top_views[['title', 'channel_title', 'views']].head(10)
```

```python
df.sort_values('likes', ascending=False)[
    ['title', 'channel_title', 'likes']
].head(10)
```

```python
df.sort_values('comment_count', ascending=False)[
    ['title', 'channel_title', 'comment_count']
].head(10)
```

```python
df['category_id'].value_counts().head(10)
```

```python
import matplotlib.pyplot as plt

df.groupby('category_id')['views'].mean().sort_values(ascending=False).plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Average Views by Category')
plt.xlabel('Category ID')
plt.ylabel('Average Views')
plt.xticks(rotation=0)
plt.show()
```

```python
daily_trending = df.groupby('trending_date').size()

daily_trending.plot(figsize=(12,5))

plt.title('Number of Trending Videos Over Time')
plt.xlabel('Trending Date')
plt.ylabel('Number of Videos')
plt.xticks(rotation=45)
plt.show()
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

def get_sentiment(text):
    score = analyzer.polarity_scores(str(text))['compound']

    if score >= 0.05:
        return 'Positive'
    elif score <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df['sentiment'] = df['title'].apply(get_sentiment)

df[['title', 'sentiment']].head(10)
```

```python
df['sentiment'].value_counts().plot(
    kind='bar',
    figsize=(7,5)
)

plt.title('Sentiment of Trending Video Titles')
plt.xlabel('Sentiment')
plt.ylabel('Number of Videos')
plt.show()
```

```python
import sqlite3

conn = sqlite3.connect('youtube_trending.db')

df.to_sql('youtube_videos', conn, if_exists='replace', index=False)

print("SQL database created")
```

```python
query = """
SELECT category_id, AVG(views) AS average_views
FROM youtube_videos
GROUP BY category_id
ORDER BY average_views DESC;
"""

sql_result = pd.read_sql(query, conn)

sql_result
```
